```python
import matplotlib.pyplot as plt
bmi=[18,20,22,24,26,28,30,32,34,36]
bp=[110,112,115,118,122,125,130,135,140,145]
plt.scatter(bmi,bp)
plt.xlabel("BMI")
plt.ylabel("BP")
plt.title("BMI vs BP")
plt.grid(True)
plt.show()
months=["Jan","Feb","March","Apr","May"]
sales=[10000,12000,15000,121212,121212]
plt.plot(months,sales,marker='o')
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Sales Analysis")
plt.grid(True)
plt.show()
departments=["cse","ece","me","civil"]
students=[10,20,30,40]
plt.bar(departments,students)
plt.xlabel("Departments")
plt.ylabel("Students")
plt.title("Department Analysis")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

departments = ["IT","HR","Finance"]

male = [120, 60, 80]

female = [90, 70, 60]

x = np.arange(len(departments))

width = 0.35

plt.bar(x-width/2, male, width, label="Male")

plt.bar(x+width/2, female, width, label="Female")

plt.xticks(x, departments)

plt.ylabel("Employees")

plt.title("Gender Distribution by Department")

plt.legend()

plt.show()
```

```python
import seaborn as sns

scores = [350,370,400,420,450,480,500,520,550,580,600]

sns.kdeplot(scores, fill=True)

plt.xlabel("Credit Score")

plt.title("KDE Plot of Credit Scores")

plt.show()
```

```python
dosage=[10,20,30,40]
effect=[5,12,18,20]
error=[1,2,2.5,3]
plt.errorbar(dosage,effect,yerr=error,fmt='o')
plt.xlabel("Dosage")
plt.ylabel("Effect")
plt.title("Dosage vs Effect")
plt.show()
```
