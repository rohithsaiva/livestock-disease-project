```python
import numpy as np
a=np.array([[1,2],
       [3,4]])
b=np.array([[5,6],
      [7,8]])
c=np.concatenate((a,b),axis=0)
d=np.concatenate((a,b),axis=1)
h=np.hstack((a,b))
v=np.vstack((a,b))
print(c)
print(d)
print(h)
print(v)
```

```python
import pandas as pd
data={
    "region":["North","South","North","East","South","East","North"],
    "sales":[200,150,300,250,180,220,275]
}
df=pd.DataFrame(data)
print(df)
region_sales=df.groupby("region")["sales"].sum()
print(region_sales)

data={
    "machine":["M1","M1","M2","M2","M3"],
    "defects":[12,15,8,10,25,]
}
df=pd.DataFrame(data)
print(df)
result=df.groupby("machine")["defects"].sum()
print(result)

data={
    "transaction":["T1","T2","T3","T4","T5"],
    "amount":[5000,7000,6500,6000,7200]
}
df=pd.DataFrame(data)
print(df)
Q1=df["amount"].quantile(0.25)
Q3=df["amount"].quantile(0.75)
IQR=Q3-Q1
lowerbound=Q1-1.5*IQR
upperbound=Q3+1.5*IQR
print(lowerbound)
print(upperbound)
outliers=df[(df["amount"]<lowerbound)|(df["amount"]>upperbound)]
print(outliers)
```
