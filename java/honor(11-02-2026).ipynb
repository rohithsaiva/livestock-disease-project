```python
# MISSING VALUES IN DATA
import pandas as pd
import numpy as np
data={
    "NAME": ["ASHA","RAVI","John"],
    "AGE": [12, np.nan, 22],
    "SCORE": [81, 35, np.nan]
}
df=pd.DataFrame(data)
print("Original DataFrame:\n")
print(df)
print("Miissing values of column:\n")
print(df.isnull().sum())
```

```python
# Handling missing values
import pandas as pd
import numpy as np
data={
    "Patient": ["Asha","Ravi","John"],
    "BP": [120, np.nan, 130]
}
df=pd.DataFrame(data)
print("Original DataFrame:\n")
print(df)
df_constant=df.copy()
df_constant["BP"]=df_constant["BP"].fillna(0)
df_mean=df.copy()
df_mean["BP"]=df_mean["BP"].fillna(df_mean["BP"].mean())
df_ffill=df.copy()
df_ffill["BP"]=df_ffill["BP"].ffill()
df_drop=df.dropna()
print("\nDataFrame with constant value:\n")
print(df_constant)
print("\n After Mean Replacement:\n")
print(df_mean)
print("\n After Forward Fill:\n")
print(df_ffill)
```
