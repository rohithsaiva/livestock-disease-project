```python
import numpy as np
a= np.arange(12)
print (a)
# reshape
print("\n reshaped array(4*3):")
print(a.reshape(4,3))
#flatten
print("\n flatten array:")
print(a.reshape(4,3).flatten())
#change data type
print("\n array after astype(float):")
print(a.astype(float))
#transpose
print("\n transpose of 4*3 array:")
print(a.reshape(4,3).T)
```

```python
import numpy as np
G = np.array(([1,2],[3,4]))
T = np.array(([5,6],[7,8]))
print(G)
print(T)
print(np.concatenate((G,T),axis = 0))
print(np.concatenate((G,T),axis = 1))
print(np.hstack((G,T)))
print(np.vstack((G,T)))
```

```python
import numpy as np
v1=np.array([1,2])
v2=np.array([3,4])
print(np.dot(v1,v2))
A=np.array([[1,2],
            [3,4]])
B=np.array([[5,6],
            [7,8]])
print(A@B)
print(A.dot(B))
print(np.linalg.det(A))
print(np.linalg.inv(A))
```
