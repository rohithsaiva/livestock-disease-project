```python
import matplotlib.pyplot as plt
sales=[50,60,70,80]
profit=[5,7,9,10]
share=[200,400,600,800]
plt.scatter(sales,profit,s=share,alpha=0.5)
plt.xlabel('sales')
plt.ylabel('profit')
plt.title('sales vs profit')
plt.show()
```

```python
labels=["Airtel","Jio","VI","BSNL"]
Sizes=[35,40,15,10]
plt.pie(Sizes,labels=labels,autopct='%1.1f%%')
plt.title("Telecom market share")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
df=pd.DataFrame({
    "Age":[30,40,50,60],
    "BMI":[22,25,28,30],
    "BP":[120,130,140,150]
})
sns.heatmap(df.corr(),annot=True,cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
import seaborn as sb
import pandas as pd
import matplotlib.pyplot as plt
data = {"social":[36,58,96],
        "maths":[56,76,89],
        "science":[90,20,100]}
df = pd.DataFrame(data)
corr = df.corr()
sb.heatmap(corr,annot = True,cmap="coolwarm")
plt.title("Correlation Plot")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
x=[1,2,3,4,5,6,7]
y=[2,4,5,4,7,8,9] # Added one element to match the length of x
sns.regplot(x=x,y=y)
plt.xlabel("x value")
plt.ylabel("y value")
plt.title("regression plot")
plt.show()
```

```python
from pywaffle import Waffle
import matplotlib.pyplot as plt
data={'a':40,'b':30,'c':20,'d':10}
fig = plt.figure(
    FigureClass=Waffle,
    rows=5,
    values=data,
    title={'label':'Waffle Chart'} # Corrected 'lable' to 'label'
)

plt.show()
```
