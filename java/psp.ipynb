```python
import numpy as np
import statistics as stats

data = [10, 20, 30, 40, 50]

mean = np.mean(data)
median = np.median(data)
variance = np.var(data)
mode = stats.mode(data)
sample_variance = np.var(data, ddof=1)

print("Mode:", mode)
print("Mean:", mean)
print("Median:", median)
print("Variance:", variance)
print("Sample Variance:", sample_variance)
```
